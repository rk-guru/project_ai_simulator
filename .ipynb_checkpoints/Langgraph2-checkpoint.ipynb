```python
""" simple Node graph """

from typing import Dict , TypedDict
from langgraph.graph import StateGraph

class AgentState(TypedDict):
    message :str

def greeting_node(state: AgentState) -> AgentState:
    print(state)
    state['message'] ='Hey'+state['message']+', how are you doing ?'
    return state

def greet_slap(state: AgentState) -> AgentState:
    state['message'] ='Hey'+state['message']+',go out'
    return state
```

```python
graph1=StateGraph(AgentState)

graph1.add_node('greeter',greeting_node)
graph1.add_node('greeter2',greet_slap)
```

```python
graph1.set_entry_point('greeter')
graph1.set_finish_point('greeter2')

app=graph1.compile()
```

```python
from typing import TypedDict, Annotated
from langgraph.graph import StateGraph, START, END
from langchain_community.chat_models import ChatOllama
from langchain_core.messages import BaseMessage, HumanMessage, AIMessage
```

```python
class AgentState(TypedDict):
    oper:str
    num1:int
    num2:int
    final:int

def addd(state:AgentState) ->AgentState:
    state['final']=state['num1']+state['num2']
    return state

def subb(state:AgentState) ->AgentState:
    state['final']=state['num1']-state['num2']
    return state

def decide(state:AgentState)->AgentState:
    if state['oper'] =='+':
        return 'add_op'
    if state['oper']=='-':
        return 'sub_op'

graph=StateGraph(AgentState)

graph.add_node('adddd',addd)
graph.add_node('subbb',subb)
graph.add_node('rout',lambda state:state)

graph.add_edge(START,'rout')

graph.add_conditional_edges('rout',decide,
                           {
                               'add_op':'adddd',
                               'sub_op':'subbb'
                           })

graph.add_edge('adddd',END)
graph.add_edge('subbb',END)

app=graph.compile()
```

```python
import operator
from typing import TypedDict, Sequence, Annotated

# Import necessary components from langchain and langgraph
from langchain_core.messages import BaseMessage, HumanMessage, AIMessage, SystemMessage
from langchain_community.chat_models import ChatOllama
from langchain_core.tools import tool
from langgraph.graph import StateGraph, END, START
from langgraph.prebuilt import ToolNode

# 1. Define the AgentState
# The 'messages' key will accumulate BaseMessage objects.
class AgentState(TypedDict):
    messages: Annotated[Sequence[BaseMessage], operator.add]
    # We'll add a 'context' field to demonstrate injecting arbitrary data
    context: Annotated[str | None, operator.setitem] # Use setitem to replace context

# 2. Define a simple tool
@tool
def multiply(a: int, b: int) -> int:
    """Multiplies two integers together."""
    return a * b

tools = [multiply]

# 3. Initialize the Language Model
# Ensure Ollama is running and 'llama3.2' model is available
try:
    llm = ChatOllama(model='llama3.2', temperature=0.7).bind_tools(tools)
    model = llm
except Exception as e:
    print(f"Could not connect to Ollama or model not found: {e}")
    print("Falling back to a dummy model for demonstration purposes.")
    # Fallback for demonstration if Ollama isn't set up
    from langchain_core.runnables import RunnableLambda
    class DummyAIMessage(AIMessage):
        tool_calls = [] # Dummy for demonstration without actual tool calls

    def dummy_invoke(messages):
        # Simulate an LLM response based on user input
        last_user_message = next((m for m in reversed(messages) if isinstance(m, HumanMessage)), None)
        if last_user_message and "multiply" in last_user_message.content.lower():
            # Simulate a tool call if 'multiply' is mentioned
            return DummyAIMessage(
                content="Okay, let me calculate that for you.",
                tool_calls=[{"name": "multiply", "args": {"a": 5, "b": 3}}]
            )
        return DummyAIMessage(content=f"Hello, I received your message: '{last_user_message.content if last_user_message else 'None'}'. How can I help?")

    model = RunnableLambda(dummy_invoke).bind_tools(tools)


# 4. Define Graph Nodes

# Node for LLM interaction
def call_model(state: AgentState) -> AgentState:
    """Calls the LLM with the current messages and returns its response."""
    print(f"\n--- call_model node ---")
    print(f"Current messages before LLM call: {[m.content for m in state['messages']]}")
    print(f"Current context: {state.get('context', 'No context')}")

    # Add a system prompt for context
    system_prompt = SystemMessage(content="You are a helpful assistant. Use tools when appropriate.")
    messages_to_send = [system_prompt] + state['messages']

    response = model.invoke(messages_to_send)
    print(f"LLM response: {response.content}")
    return {"messages": [response]}

# Conditional edge logic
def should_continue(state: AgentState) -> str:
    """
    Determines whether to continue to tools or end the conversation.
    If the last message contains tool calls, continue; otherwise, end.
    """
    last_message = state['messages'][-1]
    if last_message.tool_calls:
        print(f"--- should_continue: detected tool calls, continuing to 'tools' ---")
        return "continue"
    else:
        print(f"--- should_continue: no tool calls, ending ---")
        return "end"

# 5. Build the LangGraph
graph = StateGraph(AgentState)

graph.add_node("llm_node", call_model)
tool_node = ToolNode(tools=tools)
graph.add_node("tool_executor", tool_node)

# Define the workflow
graph.add_edge(START, "llm_node")
graph.add_conditional_edges(
    "llm_node",
    should_continue,
    {"continue": "tool_executor", "end": END}
)
graph.add_edge("tool_executor", "llm_node") # After tool execution, go back to LLM for summary

app = graph.compile()

# Helper function to print the stream
def print_stream(stream):
    print("\n--- Stream Output ---")
    for s in stream:
        print(s)
        print("-" * 20)
    print("--- End Stream Output ---\n")

# --- Demonstrating injectState ---

print("--- Scenario 1: Starting fresh (no injectState) ---")
initial_input = {"messages": [HumanMessage(content="What is 7 + 8?")]}
stream1 = app.stream(initial_input, stream_mode="values")
print_stream(stream1)


print("\n--- Scenario 2: Starting with pre-defined messages using injectState ---")
# Here, we're providing an initial state with existing messages.
# The graph will start as if these messages were already exchanged.
pre_existing_messages = [
    HumanMessage(content="Hello there!"),
    AIMessage(content="Hi! How can I assist you today?"),
    HumanMessage(content="I need to know what 12 multiplied by 5 is.")
]
# injectState is typically used with `.invoke()` or `.stream()` as an argument
# or can be used with `.get_graph().invoke/stream` if you have a compiled graph
# to set the initial state.
# For app.stream, you can pass the initial state directly in the inputs dict
# This is functionally similar to injectState when starting from START.
stream2_input = {
    "messages": pre_existing_messages
}
stream2 = app.stream(stream2_input, stream_mode="values")
print_stream(stream2)


print("\n--- Scenario 3: Simulating resuming a conversation (injecting state mid-way) ---")
# Imagine we saved the state after the LLM made a tool call and before the tool executed.
# Now we want to resume and let the tool execute.
# For simplicity, we'll manually construct a state that looks like the graph
# just finished `llm_node` and is about to enter `tool_executor`.
# In a real scenario, you'd save the actual state from a previous run.
saved_partial_state = {
    "messages": [
        HumanMessage(content="Please multiply 10 by 20."),
        AIMessage(content="Okay, I can do that.", tool_calls=[{"name": "multiply", "args": {"a": 10, "b": 20}}])
    ],
    "context": "Resumed from a checkpoint after LLM decided to use a tool."
}

# When resuming from a specific node, you use the 'from' argument.
# When `from` is specified, `injectState` essentially acts as the starting point,
# overriding the default START node and starting execution from the specified node
# with the provided state.
# Note: For `app.stream`, you'd typically pass the state as the *initial* input.
# The concept of `injectState` is more explicit when building dynamic graphs
# or resuming from exact states in the state machine using `get_graph().invoke(input, config={'recursion_limit': N}, state=saved_state)`.
# For the `app.stream` method of a compiled graph, injecting a specific 'initial'
# state means simply providing it in the input dictionary. LangGraph will ensure
# the state is correctly processed from the START node.

print("This demonstration directly sets the input as a 'resumed' state for stream.")
print("In a real application, you'd load a complete state object and use it with app.invoke/stream directly.")

stream3_input = saved_partial_state # This acts as the injected state for a new stream
stream3 = app.stream(stream3_input, stream_mode="values")
print_stream(stream3)

print("\n--- Scenario 4: Injecting only context ---")
# Start a new conversation, but provide some background context directly via state.
stream4_input = {
    "messages": [HumanMessage(content="What's the capital of France?")],
    "context": "This user is interested in geography."
}
stream4 = app.stream(stream4_input, stream_mode="values")
print_stream(stream4)

```

```python
from typing import Annotated , Sequence ,TypedDict
from langchain_core.messages import BaseMessage ,ToolMessage ,SystemMessage
from langgraph.graph.message import add_messages
from langgraph.graph import StateGraph  ,END ,START
from langgraph.prebuilt import ToolNode
from langchain.chat_models import init_chat_model
```

```python
document_content=''

class AgentState(TypedDict):
    message:Annotated[Sequence[BaseMessage],add_messages]

@tool
def update(content :str)->str:
    "update the document with the provided content"
    global document_content
    document_content=content
    return f'Document has been updated successfully ! the current content is :\n{document_content}'

@tool
def save(filename:str):
    """Save the current document to a text file and finish the process
    Args:
        filename: Name for the text file
        """

    global document_content
    if not filename.endswith('.txt'):
        filename=f'{filename}.txt'

    try:
        with open(filename,'w') as file:
            file.write(document_content)
        print(f'\n Document has been saved to :{filename}')
        return f'Document has been saved successfully to {filename}'

    except Exception as e:
        return f'Error saving docuemnt :{str(e)}'

tools=[update,save]

llm=init_chat_model(model='llama3.2',model_provider="ollama",  # Explicitly state the provider
    temperature=0.7,).bind_tools(tools)

def our_agent(state:AgentState) -> AgentState:
    system_prompt =SystemMessage(content=f'''
    You are Drafter, a helpful writing assistant. You are going to help the user update and
        - If the user wants to update or modify content, use the 'update' tool with the complete
        - If the user wants to save and finish, you need to use the 'save' tool.
        - Make sure to always show the current document state after modifications.
The current document content is: {document_content}
    ''')
    if not state['message']:
        user_input='Im ready to help you update a document .what would you like to create ?'
        user_message=HumanMessage(content=user_input)
    else:
        user_input=input(' what would you like to do with the document ?')
        user_message=HumanMessage(content=user_input)

    all_message=[system_prompt] + list(state['message']) +[user_message]
    response=llm.invoke(all_message)

    print(response.content)

    if hasattr(response, "tool_calls") and response.tool_calls:
        print(f"USING TOOLS:{[tc[ 'name'] for tc in response.tool_calls]}")
    
    return {'message':list(state['message']) +[user_message,response]}

def should_continue(state:AgentState)->str:
    'Determine if we should continue or end the conversation'
    message =state['message']
    if not message:
        return 'continue'
    for message in reversed(message):
        if (isinstance(message,ToolMessage) and 
           'saved' in message.content.lower() and 'document' in message.content.lower()):
            return 'end'

    return 'continue'

def print_message(message):
    'Function I made to print the messages in a more readable format'
    if not messages:
        return
    for message in messages[-3:]:
        if isinstance(message, ToolMessage) :
            print(f"TOOL RESULT: {message. content}" )

graph=StateGraph(AgentState)
graph.add_node('agent',our_agent)
graph.add_node('tools',ToolNode(tools))
graph.add_edge(START,'agent')  
graph.add_edge('agent','tools')
graph.add_conditional_edges('tools',should_continue,{
    'continue':'agent',
    'end':END
})

app=graph.compile()

def run_document_agent():
    state={'messages':[]}
    for step in app.stream(state,stream_mode='values'):
        if 'messages' in step:
            print_message(step['message'])

# if __name__=='main':
run_document_agent()
```

```python
import operator
from typing import TypedDict, Sequence, Annotated

# 1. FIX: Added all necessary imports
from langchain_core.messages import BaseMessage, HumanMessage, AIMessage, SystemMessage, ToolMessage
from langchain_community.chat_models import ChatOllama # Assuming init_chat_model refers to ChatOllama
from langchain_core.tools import tool
from langgraph.graph import StateGraph, START, END
from langgraph.prebuilt import ToolNode

# Global variable to store document content
document_content = ''

# 2. FIX: Renamed 'message' to 'messages' and used operator.add
class AgentState(TypedDict):
    messages: Annotated[Sequence[BaseMessage], operator.add]

@tool
def update(content: str) -> str:
    """Update the document with the provided content.
    Args:
        content: The complete new content for the document.
    """
    global document_content
    document_content = content
    return f'Document has been updated successfully! The current content is:\n{document_content}'

@tool
def save(filename: str) -> str:
    """Save the current document to a text file and finish the process.
    Args:
        filename: Name for the text file.
    """
    global document_content
    if not filename.endswith('.txt'):
        filename = f'{filename}.txt'

    try:
        with open(filename, 'w') as file:
            file.write(document_content)
        print(f'\nDocument has been saved to: {filename}')
        return f'Document has been saved successfully to {filename}'
    except Exception as e:
        return f'Error saving document: {str(e)}'

tools = [update, save]

# Initialize the chat model
# FIX: Using ChatOllama directly. Added a fallback for demonstration if Ollama isn't available.
try:
    # llm = ChatOllama(
    #     model='llama3.2',
    #     temperature=0.7,
    # ).bind_tools(tools)
    llm=init_chat_model(model='llama3.2',model_provider="ollama",  # Explicitly state the provider
    temperature=0).bind_tools(tools)
except Exception as e:
    print(f"Error initializing ChatOllama: {e}")
    print("Please ensure Ollama is running and 'llama3.2' model is available.")
    print("Falling back to a dummy LLM for demonstration purposes.")
    from langchain_core.runnables import RunnableLambda
    
    # Dummy LLM to simulate responses and tool calls for testing purposes
    def dummy_llm_invoke(messages):
        last_user_message = next((m for m in reversed(messages) if isinstance(m, HumanMessage)), None)
        if last_user_message:
            content = last_user_message.content.lower()
            if "update content to" in content:
                # Extract content after "update content to"
                new_doc_content = content.split("update content to", 1)[1].strip().replace('"', '')
                return AIMessage(content=f"Calling update tool with: '{new_doc_content}'",
                                 tool_calls=[{'name': 'update', 'args': {'content': new_doc_content}}])
            elif "save" in content and ".txt" in content:
                filename = content.split("save", 1)[1].strip().split(" ")[0].replace('.txt','') + '.txt'
                return AIMessage(content=f"Calling save tool with: '{filename}'",
                                 tool_calls=[{'name': 'save', 'args': {'filename': filename}}])
            else:
                return AIMessage(content=f"Drafter: I received your message: '{last_user_message.content}'. "
                                        f"Current document content: '{document_content if document_content else '[empty]'}'. "
                                        f"Please tell me to 'update content to \"Your text here\"' or 'save my_document.txt'.")
        return AIMessage(content="Drafter: Ready to help. What would you like to create or update?")
    llm = RunnableLambda(dummy_llm_invoke).bind_tools(tools)


def our_agent(state: AgentState) -> AgentState:
    """
    The main agent node where the LLM processes messages and decides on actions.
    """
    global document_content # To access current document_content in system prompt

    print(f"\n--- Drafter Agent Turn ---")
    messages = state['messages'] # FIX: Corrected to 'messages'

    # 4. FIX: Removed interactive input from agent. Agent processes existing state.
    # The system prompt should reflect the current document content dynamically.
    system_prompt_content = f"""
    You are Drafter, a helpful writing assistant. Your task is to assist the user in updating and saving documents.
    Instructions:
    - If the user wants to update or modify content, use the 'update' tool. You MUST provide the complete new content for the document as the 'content' argument to the 'update' tool.
    - If the user wants to save the document and finish, you MUST use the 'save' tool with the desired filename. After saving, the process will end.
    - Always inform the user about the current document state after any modifications or before asking for the next instruction.
    - If the user just asks questions or general commands not related to update/save, respond naturally based on the current document content.
    The current document content is: {document_content if document_content else '[Document is empty]'}
    """
    system_message = SystemMessage(content=system_prompt_content)

    # Pass all relevant messages, including system prompt, to the LLM
    # Filter out previous system messages from the state to ensure the latest dynamic one is used.
    conversational_messages = [m for m in messages if isinstance(m, (HumanMessage, AIMessage, ToolMessage))]
    
    response = llm.invoke([system_message] + conversational_messages)

    print(f"LLM's thought/response: {response.content}")
    if hasattr(response, "tool_calls") and response.tool_calls:
        print(f"USING TOOLS: {[tc['name'] for tc in response.tool_calls]}")
    else:
        print(f"No tools called by LLM.")

    # 4. FIX: Return the updated state with the new AI message (and any tool calls it made)
    return {'messages': [response]}


def should_continue(state: AgentState) -> str:
    """Determine if we should continue or end the conversation based on the last message."""
    messages = state['messages'] # FIX: Corrected to 'messages'
    
    # Check if the last message is a ToolMessage indicating a successful save
    # This is the primary condition to end the graph execution.
    for msg in reversed(messages):
        if isinstance(msg, ToolMessage):
            if 'document has been saved successfully' in msg.content.lower():
                print(f"\n--- should_continue: Document saved, ending conversation ---")
                return 'end'
    
    # Check if the last AI message has tool calls.
    # If it does, the graph should proceed to execute these tools.
    last_llm_message = next((m for m in reversed(messages) if isinstance(m, AIMessage)), None)
    
    if last_llm_message and last_llm_message.tool_calls:
        print(f"--- should_continue: LLM made tool calls, transitioning to 'tools' ---")
        return 'continue_to_tools' # New state for explicit tool execution
    else:
        # If no tool calls were made by the LLM (direct response) and it's not a save confirmation,
        # the current turn of the graph is complete. The control should return to the interactive loop
        # to prompt the user for the next input. This is done by returning 'end_turn', which maps to END.
        print(f"--- should_continue: LLM responded directly or no action needed, completing current turn ---")
        return 'end_turn' # New state for ending the current graph cycle, awaiting user input


# 6. FIX: Refactored print_message to print_step_messages and corrected its logic
def print_step_messages(step_output: dict):
    """Function to print the messages from a single step in a readable format."""
    if 'messages' in step_output:
        messages_in_step = step_output['messages']
        print("\n--- Current State Messages ---")
        # Print the last few messages for context of current turn
        for message in messages_in_step[-2:]:
            if isinstance(message, HumanMessage):
                print(f"USER: {message.content}")
            elif isinstance(message, AIMessage):
                print(f"DRAETER (AI): {message.content}")
                if message.tool_calls:
                    print(f"  -> Tool Calls: {message.tool_calls}")
            elif isinstance(message, ToolMessage):
                print(f"TOOL RESULT: {message.content}")
        print("----------------------------")
    else:
        print(f"Step Output (non-message updates): {step_output}") # For debugging other state changes

# Build the LangGraph
graph = StateGraph(AgentState)

# Add nodes
graph.add_node('agent', our_agent)
graph.add_node('tools', ToolNode(tools))

# 3. FIX: Corrected Graph Flow
# START -> agent (LLM processes initial input)
graph.add_edge(START, 'agent')

# Agent conditionally decides next step based on its output (tool call or direct response)
graph.add_conditional_edges(
    'agent',         # From the 'agent' node
    should_continue, # Use this function to decide the next transition
    {
        'continue_to_tools': 'tools',  # If tool calls detected, go to 'tools'
        'end_turn': END,               # If LLM responded directly, end this graph cycle
        'end': END                     # If save message from ToolMessage, terminate entire session
    }
)

# After tools run, always go back to the agent for it to process the tool's output and respond to the user
graph.add_edge('tools', 'agent')

app = graph.compile()

# 7. FIX: Refactored run_document_agent for interactive loop
def run_document_agent():
    """Runs the interactive document agent."""
    global document_content 
    document_content = '' # Reset document content for a fresh run at the start of a new session

    print("Welcome to Drafter, your writing assistant!")
    print(f"Initial document content: {document_content if document_content else '[Document is empty]'}")
    
    # Get the initial user input to start the conversation
    initial_user_input = input("What would you like to create or update? (e.g., 'Update content to \"Hello World\"', 'Save my_document.txt')\nUSER: ")
    current_state = {"messages": [HumanMessage(content=initial_user_input)]}

    while True:
        try:
            # Stream through the graph until it reaches an END state (either 'end_turn' or 'end')
            for step_output in app.stream(current_state, stream_mode='values'):
                print_step_messages(step_output) # Print the messages from the current step
                current_state = step_output # Update the state for the next potential iteration

                # Check if the graph has explicitly reached an 'end' state due to a save command
                if current_state and current_state['messages']:
                    last_msg = current_state['messages'][-1]
                    if isinstance(last_msg, ToolMessage) and 'document has been saved successfully' in last_msg.content.lower():
                        print("\nSession explicitly ended by save command. Exiting Drafter.")
                        return # Exit the function

            # If the stream finishes and we haven't returned (i.e., it hit 'end_turn'),
            # it means the current turn is over, and we need new user input.
            user_input = input("What else would you like to do? (type 'quit' to exit)\nUSER: ")
            if user_input.lower() == 'quit':
                print("Exiting Drafter. Goodbye!")
                break
            
            # Add the new user input to the current state's messages for the next graph run
            current_state = {"messages": current_state['messages'] + [HumanMessage(content=user_input)]}

        except Exception as e:
            print(f"\nAn error occurred during agent execution: {e}")
            break # Exit loop on error

# 7. FIX: Corrected '__main__' syntax
if __name__ == '__main__':
    run_document_agent()
```
