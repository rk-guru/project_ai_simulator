````python
import pandas as pd
import os
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_community.llms import Ollama
from langchain_core.documents import Document
from langchain_core.prompts import ChatPromptTemplate # Added this import, assuming it's used later in the full script
from langchain_ollama import OllamaLLM

CSV_FILE = '2k_pdb.csv'
DB_LOCATION = "./pdb_langchain_db"
OLLAMA_EMBEDDING_MODEL = 'mxbai-embed-large'
OLLAMA_LLM_MODEL = 'qwen2.5:7b'


try:
    # --- CHANGE MADE HERE: Removed .head(50) ---
    input_data = pd.read_csv(CSV_FILE).head(100)
    input_data.fillna('is not available', inplace=True)
    names = []
    for i in input_data.columns:
        j = i.replace('_',' ')
        names.append(j)
    compl_data = [] # This variable is declared but not used. You might want to remove it if it's not needed.
    documents = []
    for i in range(len(input_data)):
        data_from_db = ''
        for j in range(2,len(names)):
            data_from_db = f'{str(data_from_db)} {str(names[j])} is {str(input_data.iloc[i,j])} .'
        print('i',i)
        doc = Document(
                page_content=data_from_db,
                metadata={
                    'rcsb id': input_data.iloc[i,0],
                    'Pubmed id':input_data.iloc[i,1]
                }
                # It's generally good practice to include an 'id' for documents if you need to reference them later
                # id=str(i) # You had this in a previous version, consider adding it back if beneficial
            )
        documents.append(doc)
    print(f"Prepared {len(documents)} documents for embedding.")


except FileNotFoundError:
    print(f"Error: '{CSV_FILE}' not found. Please ensure the CSV file is in the correct directory.")
    exit()
except Exception as e:
    print(f"Error loading CSV: {e}") # Changed from PDF to CSV
    exit()


try:
    embeddings = OllamaEmbeddings(model=OLLAMA_EMBEDDING_MODEL)
except Exception as e:
    print(f"Error initializing Ollama Embeddings: {e}")
    print(f"Please ensure Ollama is running and the model '{OLLAMA_EMBEDDING_MODEL}' is pulled (ollama pull {OLLAMA_EMBEDDING_MODEL}).")
    exit()

# --- 3. Prepare or Load Chroma Database ---
db = None
if not os.path.exists(DB_LOCATION):
    print(f"Chroma database not found at '{DB_LOCATION}'. Creating a new one...")
    if documents:
        db = Chroma.from_documents(
            documents=documents, # Use the 'documents' list
            embedding=embeddings,
            persist_directory=DB_LOCATION
        )
        # db.persist() # Chroma.from_documents with persist_directory usually handles persistence automatically on creation
        print(f"Successfully added {len(documents)} documents (from CSV) and persisted Chroma DB.") # Changed from PDF to CSV
    else:
        print("No valid documents found from CSV to add to the database. Exiting.") # Changed from PDF to CSV
        exit()
else:
    print(f"Chroma database found at '{DB_LOCATION}'. Loading existing database.")
    db = Chroma(persist_directory=DB_LOCATION, embedding_function=embeddings)
    print("Chroma database loaded successfully.")

# --- 4. Obtain the Retriever ---
if db:
    retriever = db.as_retriever()
    print("Retriever successfully created from the Chroma database!")
else:
    print("Could not create or load the Chroma database. Retriever not available.")
    exit()

# --- 5. Initialize the Language Model (LLM) ---
print(f"Initializing Ollama LLM with model: {OLLAMA_LLM_MODEL}...")
try:
    model = OllamaLLM(model=OLLAMA_LLM_MODEL) #Ollama
    print("Ollama LLM initialized successfully.")
except Exception as e:
    print(f"Error initializing Ollama LLM: {e}")
    print(f"Please ensure Ollama is running and the model '{OLLAMA_LLM_MODEL}' is pulled (ollama pull {OLLAMA_LLM_MODEL}).")
    exit()
````

````python

# --- 6. Define the Prompt Template ---
template = """
You are an expert in bioinformatics , bio technology   .

Here are some relevant protein data:
{protein_info}

Here is the question to answer: {question}
"""
prompt = ChatPromptTemplate.from_template(template)

chain = prompt | model

while True:
    question = input('Ask your question: ')
    print(f"Your question: {question}") # Changed to print only the question once

    if question.lower() == 'q':
        print("Exiting Q&A session.")
        break
    else:
        # Step 1: Retrieve relevant documents (reviews)
        # The 'review' variable will actually be a list of Document objects
        retrieved_docs = retriever.invoke(question)

        # Extract just the page_content (the review text) from the Document objects
        # and join them into a single string to pass to the prompt
        formatted_reviews = "\n\n".join([doc.page_content for doc in retrieved_docs])

        # Step 2: Invoke the chain with the retrieved reviews and the user's question
        result = chain.invoke({'reviews': formatted_reviews, 'question': question})

        # Step 3: Print the result from the LLM
        print("\n--- LLM Response ---")
        print(result) # For Ollama, the result is usually a string directly
        print("---------------------\n")
````

````python
import pandas as pd
import os
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_community.llms import Ollama
from langchain_core.documents import Document
from langchain_core.prompts import ChatPromptTemplate


CSV_FILE = '2k_pdb.csv'
DB_LOCATION = "./pdb_langchain_db"
OLLAMA_EMBEDDING_MODEL = 'mxbai-embed-large'
OLLAMA_LLM_MODEL = 'qwen2.5:7b'


try:
    input_data = pd.read_csv(CSV_FILE).head(5)
    input_data.fillna('is not available', inplace=True)
    names = []
    for i in input_data.columns:
        j = i.replace('_', ' ')
        names.append(j)
    compl_data = []
    documents = []
    for i in range(len(input_data)):
        data_from_db = ''
        for j in range(2, len(names)):
            data_from_db = f'{str(data_from_db)} {str(names[j])} is {str(input_data.iloc[i,j])} .'

        print('data_from_db',data_from_db)
        doc = Document(
                page_content=data_from_db,
                metadata={
                    'rcsb id': input_data.iloc[i,0],
                    'Pubmed id':input_data.iloc[i,1]
                },
                id=str(i)
            )
        documents.append(doc)
    print(f"Prepared {len(documents)} documents for embedding.")


except FileNotFoundError:
    print(f"Error: '{CSV_FILE}' not found. Please ensure the PDF file is in the correct directory.")
    exit()
except Exception as e:
    print(f"Error loading PDF: {e}")
    exit()


try:
    embeddings = OllamaEmbeddings(model=OLLAMA_EMBEDDING_MODEL)
except Exception as e:
    print(f"Error initializing Ollama Embeddings: {e}")
    print(f"Please ensure Ollama is running and the model '{OLLAMA_EMBEDDING_MODEL}' is pulled (ollama pull {OLLAMA_EMBEDDING_MODEL}).")
    exit()

# --- 3. Prepare or Load Chroma Database (Adjusted for PDF documents list) ---
db = None
if not os.path.exists(DB_LOCATION):
    print(f"Chroma database not found at '{DB_LOCATION}'. Creating a new one...")
    # 'documents' is already prepared from the PDF loader
    if documents:
        db = Chroma.from_documents(
            documents=documents, # Use the 'documents' list from PDF loading
            embedding=embeddings,
            persist_directory=DB_LOCATION
        )
        # db.persist()
        print(f"Successfully added {len(documents)} documents (from PDF) and persisted Chroma DB.")
    else:
        print("No valid documents found from PDF to add to the database. Exiting.")
        exit()
else:
    print(f"Chroma database found at '{DB_LOCATION}'. Loading existing database.")
    db = Chroma(persist_directory=DB_LOCATION, embedding_function=embeddings)
    print("Chroma database loaded successfully.")

# --- 4. Obtain the Retriever (Same as before) ---
if db:
    retriever = db.as_retriever()
    print("Retriever successfully created from the Chroma database!")
else:
    print("Could not create or load the Chroma database. Retriever not available.")
    exit()

# --- 5. Initialize the Language Model (LLM) (Same as before) ---
print(f"Initializing Ollama LLM with model: {OLLAMA_LLM_MODEL}...")
try:
    model = Ollama(model=OLLAMA_LLM_MODEL)
    print("Ollama LLM initialized successfully.")
except Exception as e:
    print(f"Error initializing Ollama LLM: {e}")
    print(f"Please ensure Ollama is running and the model '{OLLAMA_LLM_MODEL}' is pulled (ollama pull {OLLAMA_LLM_MODEL}).")
    exit()

# ---
# ## Corrected Prompt Template and Chain Invocation

# The key changes are here:

# ```python
# # --- 6. Define the Prompt Template ---
template = """
You are an expert in bioinformatics and biotechnology.

Here is some relevant protein data:
{protein_info}

Here is the question to answer: {question}
"""
prompt = ChatPromptTemplate.from_template(template)

chain = prompt | model

while True:
    question = input('Ask your question: ')
    print(f"Your question: {question}")

    if question.lower() == 'q':
        print("Exiting Q&A session.")
        break
    else:
        # Step 1: Retrieve relevant documents (now representing protein info)
        retrieved_docs = retriever.invoke(question)

        # Extract just the page_content (the protein data text) from the Document objects
        # and join them into a single string to pass to the prompt
        formatted_protein_info = "\n\n".join([doc.page_content for doc in retrieved_docs])

        # Step 2: Invoke the chain with the retrieved protein info and the user's question
        result = chain.invoke({'protein_info': formatted_protein_info, 'question': question})

        # Step 3: Print the result from the LLM
        print("\n--- LLM Response ---")
        print(result)
        print("---------------------\n")
````

````python
import pandas as pd
from langchain_community.document_loaders import CSVLoader
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.llms import Ollama
from langchain.chains import RetrievalQA
from langchain.prompts import PromptTemplate
import os

# --- Configuration ---
CSV_FILE_PATH = '100_data.csv' # Make sure this matches your CSV file name
OLLAMA_EMBEDDING_MODEL = 'mxbai-embed-large' # Or 'nomic-embed-text', etc.
OLLAMA_LLM_MODEL = 'llama3.2'
VECTOR_DB_PATH = './Chroma_db/chroma_db1' # Directory to store your vector database

# --- 1. Load and Process CSV Data ---
# Load the CSV file into a pandas DataFrame
df = pd.read_csv(CSV_FILE_PATH)

# Create a 'rag_content' column by combining relevant information from CSV columns.
# This concatenated string will be the primary content embedded for retrieval.
df['rag_content'] = df.apply(
    lambda row: f"rcsb_id: {row['rcsb_id']}\n"
                f"pubmed_id: {row['pubmed_id']}\n"
                f"title: {row['title']}\n"
                f"PubMed Abstract: {row['PubMed Abstract']}\n"
                f"Ligand_name: {row['Ligand_name']}",
    axis=1
)

# Convert DataFrame rows into LangChain Document objects.
# Each Document will contain the 'rag_content' and relevant metadata.
from langchain.schema import Document

documents = []
for index, row in df.iterrows():
    doc = Document(
        page_content=row['rag_content'],
        metadata={
            'source': CSV_FILE_PATH,
            'row_id': index,
            'rcsb_id': row['rcsb_id'], # Keep rcsb_id in metadata for easy access/filtering
        }
    )
    documents.append(doc)

print(f"Loaded {len(documents)} documents from the CSV.")

# --- 2. Initialize Ollama Embedding Model ---
print(f"Initializing embedding model: {OLLAMA_EMBEDDING_MODEL}...")
# This uses your local Ollama instance to generate embeddings.
embeddings = OllamaEmbeddings(model=OLLAMA_EMBEDDING_MODEL)
print("Embedding model initialized.")

vectorstore = Chroma.from_documents(
    documents=documents,
    embedding=embeddings,
    persist_directory=VECTOR_DB_PATH
)
vectorstore.persist()
llm = Ollama(model=OLLAMA_LLM_MODEL)
print("LLM ready.")

template = """Use the following context to answer the question.
If you don't know the answer, just say that you don't know, don't try to make up an answer.
You are a bioinformatics specialist and your task is to provide information about the proteins, based on the provided data.

Context:
{context}

Question: {question}
Helpful Answer:"""
QA_CHAIN_PROMPT = PromptTemplate.from_template(template)

qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=vectorstore.as_retriever(search_kwargs={"k": 5}),
    chain_type_kwargs={"prompt": QA_CHAIN_PROMPT},
    return_source_documents=True
)
print("RAG chain ready.")

# --- 6. Query the RAG System ---
print("\n--- Start Querying ---")
while True:
    query = input("Enter your query about proteins (or 'exit' to quit): ")
    if query.lower() == 'exit':
        break

    try:
        # Invoke the RAG chain with the user's query.
        response = qa_chain.invoke({"query": query})
        print("\n--- Response ---")
        print(response["result"])

        # Display the source documents that were retrieved and used to generate the answer.
        if response.get("source_documents"):
            print("\n--- Source Documents Retrieved ---")
            for i, doc in enumerate(response["source_documents"]):
                print(f"Document {i+1} (RCSB ID: {doc.metadata.get('rcsb_id', 'N/A')}):")
                # Print only the first 200 characters of the page_content for brevity
                print(doc.page_content[:200] + "...")
                print("-" * 20)

        print("----------------\n")
    except Exception as e:
        print(f"An error occurred: {e}")
        print("Please ensure Ollama server is running and models are downloaded.")
````
