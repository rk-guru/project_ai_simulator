```python
def extract_xpath_from_llm_response(llm_response_content: str) -> str:
    """
    Cleans the LLM response to remove surrounding quotes and excess whitespace, 
    ensuring only the raw XPath string is returned.
    """
    cleaned_path = llm_response_content.strip()
    if cleaned_path.startswith(('"', "'")) and cleaned_path.endswith(('"', "'")):
        cleaned_path = cleaned_path[1:-1].strip()
    return cleaned_path
```

```python
try:
    driver = webdriver.Chrome()
    driver.maximize_window()
    wait = WebDriverWait(driver, 4) 

    target_url="https://demoqa.com"
    driver.get(target_url)
    print(f"Navigated to initial URL: {driver.current_url}")
    
    ELEMENTS_GROUP_XPATH = "//h5[text()='Elements']"
    print(f"Clicking main group: 'Elements'")
    
    elements_group_tile = wait.until(
        EC.element_to_be_clickable((By.XPATH, ELEMENTS_GROUP_XPATH)))
    elements_group_tile.click()
    print("Successfully expanded 'Elements' group.")

    # 2. Prepare for the sub-link click that triggers navigation
    link_text_to_click = "Web Tables"
    WEB_TABLES_LINK_XPATH = f"//span[text()='{link_text_to_click}']"

    # --- Step A: Record the URL BEFORE the click ---
    initial_url = driver.current_url
    print(f"\nInitial URL recorded: {initial_url}")

    print(f"Proceeding to click the sub-link: '{link_text_to_click}'")
    
    link_element = wait.until(
        EC.element_to_be_clickable((By.XPATH, WEB_TABLES_LINK_XPATH))
    )
    
    # 3. Click the link to initiate navigation
    link_element.click()

    # --- Step B: Wait for the URL to change ---
    print("Waiting for URL to change...")
    # This explicit wait is the most reliable way to confirm navigation
    wait.until(EC.url_changes(initial_url))
    
    # --- Step C: Get the new URL after the change ---
    new_url = driver.current_url
    
    print(f"\nSuccessfully navigated.")
    print(f"New URL is: {new_url}")
    
    # Simple check to confirm the change
    if new_url != initial_url:
        print("SUCCESS: The URL has changed after the click.")
    else:
        print("FAILURE: The URL did not appear to change.")
        
    time.sleep(3) # Hold browser open to view

except Exception as e:
    print(f"An error occurred: {e}")
finally:
    # Close the browser session
    if 'driver' in locals():
        driver.quit()
        print("Browser closed.")
```

```python
import requests
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# --- Utility Function (Kept for context, but not essential for filling fields) ---
def extract_xpath_from_llm_response(llm_response_content: str) -> str:
    """Cleans the LLM response string."""
    cleaned_path = llm_response_content.strip()
    if cleaned_path.startswith(('"', "'")) and cleaned_path.endswith(('"', "'")):
        cleaned_path = cleaned_path[1:-1].strip()
    return cleaned_path

try:
    # --- Selenium Execution ---
    driver = webdriver.Chrome()
    driver.maximize_window()
    wait = WebDriverWait(driver, 10) 

    # We will navigate directly to the Text Box page as it has input fields.
    target_url="https://demoqa.com/text-box"
    driver.get(target_url)
    print(f"Navigated to Text Box page: {driver.current_url}")
    
    # --- Step 1: Locate and fill the Full Name field ---
    
    # The 'Full Name' input field has the ID 'userName'
    username_field_id = "userName"
    input_name = "Jane Doe"
    
    print(f"\nLocating and filling input field with ID: '{username_field_id}'")
    
    name_input = wait.until(
        EC.presence_of_element_located((By.ID, username_field_id))
    )
    
    # Use send_keys() to input the text
    name_input.send_keys(input_name)
    print(f"Entered name: {input_name}")
    
    # --- Step 2: Locate and fill the Email field ---

    # The 'Email' input field has the ID 'userEmail'
    email_field_id = "userEmail"
    input_email = "jane.doe@example.com"
    
    print(f"\nLocating and filling input field with ID: '{email_field_id}'")

    email_input = wait.until(
        EC.presence_of_element_located((By.ID, email_field_id))
    )
    
    # Use send_keys() to input the text
    email_input.send_keys(input_email)
    print(f"Entered email: {input_email}")
    
    # --- Step 3: Locate and click the Submit button ---
    
    # The Submit button has the ID 'submit'
    submit_button_id = "submit"
    
    print("\nLocating and clicking 'Submit' button...")
    
    submit_button = wait.until(
        EC.element_to_be_clickable((By.ID, submit_button_id))
    )
    
    submit_button.click()
    print("Submit button clicked.")
    
    # --- Step 4: Verify the output (optional, but good practice) ---
    
    # The submitted data appears in a div with ID 'output'
    output_element = wait.until(
        EC.presence_of_element_located((By.ID, "output"))
    )
    
    # Verify that the submitted name is visible in the output
    if input_name in output_element.text:
        print("\nSUCCESS: Data submission confirmed on the page.")
    else:
        print("\nFAILURE: Data submission output not found.")

    time.sleep(5) # Hold browser open to view

except Exception as e:
    print(f"An error occurred: {e}")
finally:
    # Close the browser session
    if 'driver' in locals():
        driver.quit()
        print("Browser closed.")
```
